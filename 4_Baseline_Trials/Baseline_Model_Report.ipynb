# Baseline Model Evaluation Report: Earthquake Magnitude Prediction

## Objective
To establish baseline performance for predicting earthquake magnitude (Mag) using only spatial and depth features (Latitude, Longitude, Depth) from the provided dataset. Three baseline models were evaluated: mean predictor, median predictor, and linear regression.

## Baseline Models Chosen
**Mean Predictor:**
Always predicts the mean magnitude from the training set. This serves as a naive baseline.

**Median Predictor:**
Always predicts the median magnitude from the training set. Robust to outliers.

**Linear Regression:**
Attempts to find a linear relationship between Latitude, Longitude, Depth and Magnitude.

## Performance Evaluation
Models were evaluated using standard regression metrics:

- **Mean Absolute Error (MAE)**: Average absolute difference between predictions and actual values.
- **Root Mean Square Error (RMSE)**: Square root of average squared difference.
- **R² Score (Linear Regression only)**: Proportion of variance in target explained by the model.

## Summary Table of Results
| Model             | MAE   | RMSE  | R²     |
|------------------|-------|-------|--------|
| Mean Predictor    | 0.347 | 0.459 | -      |
| Median Predictor  | 0.329 | 0.482 | -      |
| Linear Regression | 0.347 | 0.459 | 0.001  |

## Analysis and Insights
**Baseline Model Performance:**
Mean and median predictors performed similarly, as expected. Linear regression showed minimal improvement.

**Initial Hypotheses vs. Results:**
Expected spatial and depth features might offer predictive signal. Results show they are weak predictors.

**Model Robustness:**
Used median imputation and feature scaling. 80/20 train/test split provided fair evaluation.

## What We Learned and Next Steps
**Key Takeaways:**
- Naive predictors set a solid benchmark.
- Linear regression does not outperform significantly.
- Low R² indicates need for richer features.

**Next Steps:**
- Add features like fault type, seismic moment, historical activity.
- Explore advanced models like decision trees or neural nets.
- Apply domain knowledge for feature engineering.

## Conclusion
These baseline results provide a starting point for future modeling. They demonstrate the importance of meaningful features and guide us to enrich the dataset and model complexity in subsequent steps.
